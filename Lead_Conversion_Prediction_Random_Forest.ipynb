```python
import numpy as np
import pandas as pd
```

```python
data=pd.read_csv('/content/Unified_Leads.csv')
data
```

```python
print(data.shape)
print(data.size)
print(data.dtypes)
print(data.isnull().sum())
```

```python
data=data.drop(['Name','contact_number','Contact_email','references'],axis=1)
data
```

```python
# converting columns creation_date,first_call _attempt,last_call_date to datetime
data['creation_date']=pd.to_datetime(data['creation_date'],errors='coerce',format='mixed')
data['first_call_attempt']=pd.to_datetime(data['first_call_attempt'],errors='coerce',format='mixed')
data['last_call_date']=pd.to_datetime(data['last_call_date'],errors='coerce',format='mixed')
```

```python
# creating new columns
data['cr_day']=data['creation_date'].dt.day
data['cr_month']=data['creation_date'].dt.month
data['cr_year']=data['creation_date'].dt.year
data['cr_hour']=data['creation_date'].dt.hour
data['cr_minute']=data['creation_date'].dt.minute
```

```python
data['fst_day']=data['first_call_attempt'].dt.day.fillna(-1).astype('int64')
data['fst_month']=data['first_call_attempt'].dt.month.fillna(-1).astype('int64')
data['fst_year']=data['first_call_attempt'].dt.year.fillna(-1).astype('int64')
data['fst_hour']=data['first_call_attempt'].dt.hour.fillna(-1).astype('int64')
data['fst_minute']=data['first_call_attempt'].dt.minute.fillna(-1).astype('int64')
```

```python
data['lst_day']=data['last_call_date'].dt.day.fillna(-1).astype('int64')
data['lst_month']=data['last_call_date'].dt.month.fillna(-1).astype('int64')
data['lst_year']=data['last_call_date'].dt.year.fillna(-1).astype('int64')
data['lst_hour']=data['last_call_date'].dt.hour.fillna(-1).astype('int64')
data['lst_minute']=data['last_call_date'].dt.minute.fillna(-1).astype('int64')
```

```python
print(data.isnull().sum())
```

```python
data['lead_tag']=data['lead_tag'].fillna('no tag')
```

```python
# dropping the columns creation_date,first_call_attempt,last_call_date,last_call_time bcz relevant measures are already created from this
data=data.drop(['creation_date','first_call_attempt','last_call_date','last_call_time'],axis=1)
```

```python
data
```

```python
print(data.dtypes)
```

```python
print(data.isnull().sum())
```

```python
# encoding categorical columns
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
data['lead_tag']=le.fit_transform(data['lead_tag'])
data['lead_status']=le.fit_transform(data['lead_status'])
data['assigned_to']=le.fit_transform(data['assigned_to'])
data['campaign_name']=le.fit_transform(data['campaign_name'])
data['source_name']=le.fit_transform(data['source_name'])
data['source_type']=le.fit_transform(data['source_type'])
```

```python
# since our target column is lead_stage mapping the value converted as 1 and all others as 0 bcz the task is to perform a binary classification
data['lead_stage']=data['lead_stage'].apply(lambda x:1 if str(x).lower()=='converted' else 0)
```

```python
data
```

```python
# assigning lead_stage to target variable and dropping the original lead_stage column
data['target']=data['lead_stage']
```

```python
data=data.drop(['lead_stage'],axis=1)
```

```python
data
```

```python
# splitting into x and y
x=data.iloc[:,:-1].values
x
```

```python
y=data.iloc[:,-1].values
y
```

```python
# splitting into training and testing
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.20)
```

```python
# preprocessing
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x_train=sc.fit_transform(x_train)
x_test=sc.transform(x_test)
```

```python
x_train
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
from sklearn.ensemble import RandomForestClassifier
rf=RandomForestClassifier()
rf.fit(x_train,y_train)
y_pred=(rf.predict(x_test))
print(accuracy_score(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred))
```
